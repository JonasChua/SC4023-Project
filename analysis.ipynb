```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean, professional style for the notebook plots
sns.set_theme(style="whitegrid")

# Load the CSV files generated by your main script
df_raw = pd.read_csv("query_results_raw.csv")
df_avg = pd.read_csv("query_results_average.csv")

print("Data loaded successfully!")
```

```python
# Display the first 5 rows for each configuration type
for q_type in df_raw['type'].unique():
    print(f"\n--- {q_type} (First 5 Runs) ---")
    display(df_raw[df_raw['type'] == q_type].head(5))
```

```python
print("--- Average Results (Over 50 Iterations) ---")
display(df_avg)
```

```python
plt.figure(figsize=(10, 6))

# 1. Draw the boxplot, but make the boxes lighter/thinner
sns.boxplot(
    data=df_raw, 
    x='type', 
    y='phases.Query Execution', 
    width=0.4,               # Make boxes narrower
    color='lightgray',       # Neutral background for the boxes
    showfliers=False         # Hide the default outlier circles so they don't double-print
)

# 2. Overlay the actual 50 data points with a "stripplot"
sns.stripplot(
    data=df_raw, 
    x='type', 
    y='phases.Query Execution', 
    palette='viridis',       # Add a nice color gradient
    size=5,                  # Size of the dots
    alpha=0.6,               # Make dots slightly transparent so overlaps are visible
    jitter=True              # Spread them out horizontally so they don't form a single line
)

plt.title('Query Execution Time by Configuration (50 Iterations)', fontsize=14, fontweight='bold')
plt.ylabel('Execution Time (seconds)', fontsize=12)
plt.xlabel('Database Configuration', fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# A barplot in seaborn automatically calculates the mean and adds an error bar 
# (representing a 95% confidence interval or standard deviation)
sns.barplot(
    data=df_raw, 
    x='type', 
    y='phases.Query Execution', 
    palette='magma',     # Use a distinct color palette
    capsize=0.1,         # Adds little horizontal caps to the error bars
    errorbar='sd'        # Show standard deviation
)

plt.title('Average Query Execution Time by Configuration', fontsize=14, fontweight='bold')
plt.ylabel('Execution Time (seconds)', fontsize=12)
plt.xlabel('Database Configuration', fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Load the average results
df_avg = pd.read_csv("query_results_average.csv")

# Sort by execution time so the slowest is at the top, fastest at the bottom
df_avg_sorted = df_avg.sort_values('phases.Query Execution', ascending=False)

# Select only the columns we really care about for a clean presentation
columns_to_show = [
    'type', 
    'phases.Query Execution', 
    'phases.Initialisation', 
    'totals.read'
]
df_clean = df_avg_sorted[columns_to_show]

# Apply styling to create a color-coded table
styled_table = (
    df_clean.style
    .background_gradient(cmap='Reds', subset=['phases.Query Execution', 'totals.read'])
    .format({
        'phases.Query Execution': '{:.4f} sec',
        'phases.Initialisation': '{:.4f} sec',
        'totals.read': '{:,.0f} blocks'
    })
    .set_caption("Average Database Query Performance (Over 50 Runs)")
    .set_table_styles([{
        'selector': 'caption',
        'props': [('font-size', '16px'), ('font-weight', 'bold'), ('color', 'black'), ('margin-bottom', '10px')]
    }, {
        'selector': 'th',
        'props': [('text-align', 'left'), ('background-color', '#f4f4f4')]
    }])
    .hide(axis="index") # Hides the arbitrary row numbers on the left
)

# Display the beautiful HTML table
display(styled_table)
```

```python
import pandas as pd

# Load the raw data (we need the raw data, not the average, to calculate standard deviation)
df_raw = pd.read_csv("query_results_raw.csv")

# Group by 'type' and calculate comprehensive statistics for Query Execution
# .describe() automatically calculates count, mean, std, min, 25%, 50%, 75%, max
stats_df = df_raw.groupby('type')['phases.Query Execution'].describe()

# Sort by the mean execution time so it flows from slowest to fastest
stats_df = stats_df.sort_values('mean', ascending=False)

# Let's style this table so it looks professional for your report
styled_stats = (
    stats_df.style
    # Highlight the standard deviation in a subtle red so the variance stands out
    .background_gradient(cmap='Reds', subset=['std']) 
    # Highlight the mean in blue
    .background_gradient(cmap='Blues', subset=['mean'])
    # Format everything to 4 decimal places since your variance is likely tiny
    .format("{:.4f}")
    # Format the 'count' column back to a clean integer (50)
    .format("{:.0f}", subset=['count'])
    .set_caption("Detailed Statistical Analysis: Query Execution Time (Seconds)")
    .set_table_styles([{
        'selector': 'caption',
        'props': [('font-size', '16px'), ('font-weight', 'bold'), ('color', 'black'), ('margin-bottom', '10px')]
    }, {
        'selector': 'th',
        'props': [('text-align', 'left'), ('background-color', '#f4f4f4')]
    }])
)

# Render the table
display(styled_stats)
```

```python
plt.figure(figsize=(10, 6))

# Sort the values from highest to lowest for a cleaner visual
df_avg_sorted = df_avg.sort_values('totals.read', ascending=False)

# Bar plot for total blocks read
sns.barplot(data=df_avg_sorted, x='type', y='totals.read', hue='type', palette='viridis', legend=False)

plt.title('Average Total Blocks Read by Configuration', fontsize=14, fontweight='bold')
plt.ylabel('Blocks Read', fontsize=12)
plt.xlabel('Database Configuration', fontsize=12)
plt.xticks(rotation=45)

# Render the plot directly in the notebook
plt.show()
```
